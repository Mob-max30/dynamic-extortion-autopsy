# Basic Statistical Analysis
**Expected Input:** Dataset containing numerical attributes.
**Expected Output:** Statistical summary showing central tendency and dispersion measures.
**Learning Outcome:** Ability to perform basic statistical analysis and interpret numerical data characteristics.

```
import pandas as pd
import numpy as np

# Load dataset and upgrade data types
df = pd.read_csv('Bangalore_Ride_Forensics_Clean.csv')
df = df.convert_dtypes()

# Display the first two rows
df.head(2)
```

### 1. Extract Numerical Attributes
Using `select_dtypes` to dynamically isolate numerical columns.

```
num_df = df.select_dtypes(include=['number'])
num_df.head(2)
```

### 2. Manual Statistical Summary (Handling Modes)
Calculating individual measures. The use of `.iloc[0]` on the mode extracts the primary value in case of ties. These metrics are then compiled into a structured DataFrame.

```
mean_vals = num_df.mean()
median_vals = num_df.median()
mode_vals = num_df.mode().iloc[0]
std_vals = num_df.std()
min_vals = num_df.min()
max_vals = num_df.max()

manual_summary = pd.DataFrame({
    'Mean': mean_vals,
    'Median': median_vals,
    'Mode': mode_vals,
    'Std_Deviation': std_vals,
    'Minimum': min_vals,
    'Maximum': max_vals
})

# Display the summary table
manual_summary
```

### 3. Automated Statistical Summary
The built-in method computes count, mean, std, min, max, and percentiles automatically.

```
automated_summary = num_df.describe()
automated_summary
```
