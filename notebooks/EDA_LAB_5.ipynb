```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/content/Bangalore_Ride_Forensics_Duplicates.csv")
```

```python
dup_mask = df.duplicated(keep= False)
df.duplicated().sum()
```

```python
df[dup_mask].sort_values("ID")
dup_ids = df[df["ID"].duplicated(keep=False)].sort_values("ID")
dup_ids
```

```python
for ride_id, group in dup_ids.groupby('ID'):
    print(f"\n ride id is: {ride_id}")
    display(group)
```
