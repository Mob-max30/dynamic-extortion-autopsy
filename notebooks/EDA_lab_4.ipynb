```python
mapping = {
    "malleshwaram": "Malleshwaram", "MALLESHWARAM": "Malleshwaram", "Malleswaram": "Malleshwaram", "Malleshwaram ": "Malleshwaram",
    "marathahalli": "Marathahalli", "MARATHAHALLI": "Marathahalli", "Marathalli": "Marathahalli", "Marathahalli ": "Marathahalli",
    "HSR": "HSR Layout", "hsr layout": "HSR Layout", "HSR  Layout": "HSR Layout", "H.S.R Layout": "HSR Layout", "HSR Layout ": "HSR Layout",
    "bellandur": "Bellandur", "BELLANDUR": "Bellandur", "Bellandoor": "Bellandur", " Bellandur": "Bellandur",
    "indiranagar": "Indiranagar", "INDIRANAGAR": "Indiranagar", "Indira Nagar": "Indiranagar", "Indiranagar ": "Indiranagar",
    "whitefield": "Whitefield", "WHITEFIELD": "Whitefield", "White Field": "Whitefield", "Whitefield ": "Whitefield",
    "electronic city": "Electronic City", "ELECTRONIC CITY": "Electronic City", "E-City": "Electronic City", "Electronic  City": "Electronic City", "Electronic City ": "Electronic City",
    "peenya": "Peenya", "PEENYA": "Peenya", "Penya": "Peenya", "Peenya ": "Peenya",
    "koramangala": "Koramangala", "KORAMANGALA": "Koramangala", "Koramangla": "Koramangala", "Koramangala ": "Koramangala",
    "hebbal": "Hebbal", "HEBBAL": "Hebbal", "Hebal": "Hebbal", " Hebbal": "Hebbal",
}
```

```python
clean_df['Pickup'] = clean_df['Pickup'].replace(mapping)
clean_df['Dropoff'] = clean_df['Dropoff'].replace(mapping)
```

```python
clean_df = clean_df.dropna(subset=["ID", "Time"])
```

```python
num_cols = clean_df.select_dtypes("number").columns
for col in num_cols:
    med = clean_df[col].median()
    clean_df[col] = clean_df[col].fillna(med)
    print(f"{col}: median used = {med:.2f}")
```

```python
cat_cols = clean_df.select_dtypes(exclude="number").columns.drop(["ID", "Time"])
for col in cat_cols:
    mode = clean_df[col].mode()[0]
    clean_df[col] = clean_df[col].fillna(mode)
    print(f"{col}: mode used = {mode}")
```

```python
cols = list(num_cols)

comparison = pd.DataFrame({
    "Raw_Mean":       df[cols].mean(),
    "Cleaned_Mean":   clean_df[cols].mean(),
    "Raw_Median":     df[cols].median(),
    "Cleaned_Median": clean_df[cols].median(),
    "Raw_Mode":       df[cols].mode().iloc[0],
    "Cleaned_Mode":   clean_df[cols].mode().iloc[0],
}).round(2)
comparison
```
